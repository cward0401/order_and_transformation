# Harmonic Structure of an Oscillatory Signal

## Objective
Examine the harmonic structure of a sythetic oscillarotry signal using spectral analysis. 

This notebokk studies how periodic behavior appears in the frequency domain and compares the spectral representation of the true signal with that of noisy observations. The goal is to connect time-doain oscillation with its corresponding frequency-domain signature.

-------------------------------------------------------------------------------------------------------------------------------------------------------

## Method Overview
This notebook proceeds in three stages:
1. Recreate the synthetic osillatory system.
2. Compute the power spectrum of the signal
3. Compare the spectral structure of the true signal and noisy observations.

This provides a frequency-domain perspective on the same system analyzed previously through state estimation, residual diagnostics, and phase-space geometry.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import periodogram

np.random.seed(72)
```

### Recreate the Synthetic Signal

```python
t = np.arange(0, 200)

true_signal = np.sin(2*np.pi*t/30)

noise = np.random.normal(0, 0.4, len(t))

observations = true_signal + noise
```

```python
plt.figure(figsize=(10,4))
plt.plot(t, true_signal, label="True Signal")
plt.scatter(t, observations, s=12, label="Observed Measurements")
plt.title("Synthetic Oscillatory Signal")
plt.xlabel("Time Step")
plt.ylabel("Signal Value")
plt.legend()
plt.show()
```

### Compute the Spectrum of the True Signal
To examine the harmonic structure of the system, we compute the *power spectrum* of the true signal.

The power spectrum shows how the signal's variance is dristributed across frequencies. For a periodic signal, we expect to see a dominant peak at the frequency corresponding to its oscillation period.

```python
freqs_true, power_true = periodogram(true_signal)
```

```python
plt.figure(figsize=(10,4))
plt.plot(freqs_true, power_true)

plt.title("Power Spectrum of the True Signal")
plt.xlabel("Frequency")
plt.ylabel("Power")
plt.show()
```

### Compute the Spectrum of the Observed Signal
Follwing is a power spectrum computation of the noisy observations.

This allows us to compare the clean harmonic structure of the true signal with the frequency-domain representation of the measured data.

```python
freq_obs, power_obs = periodogram(observations)
```

```python
plt.figure(figsize=(10,4))
plt.plot(freq_obs, power_obs)

plt.title("Power Spectrum of the Observed Signal")
plt.xlabel("Frequency")
plt.ylabel("Power")

plt.show()
```

### Interpretation
Ther tue signal spevtrum contains a single dominant peak at the expected oscillation frequency, reflecting the simple harmonic structure of the system.

The observed signal retains this dominant peak, but measurement noise introduces additional pwer across a wider range of frequencies. As a result, the observed spectrum is less clean, though the primary periodic component remains clearly identifiable. 

This comparison shows that spectral analysis can recover the dominant frequency structure of a system even when the observations are noisy.
